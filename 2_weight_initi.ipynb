```python
import tensorflow as tf
from tensorflow.keras import layers, models, initializers

# Load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

results = {}

# -------- Random Initialization --------
print("\nRandom Initialization")

model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu', kernel_initializer='random_normal'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x_train, y_train, epochs=3, verbose=0)

loss, acc = model.evaluate(x_test, y_test, verbose=0)
results['Random'] = acc
print("Accuracy:", acc)


# -------- Zeros Initialization --------
print("\nZeros Initialization")

model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu', kernel_initializer='zeros'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x_train, y_train, epochs=3, verbose=0)

loss, acc = model.evaluate(x_test, y_test, verbose=0)
results['Zeros'] = acc
print("Accuracy:", acc)


# -------- He Initialization --------
print("\nHe Initialization")

model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu', kernel_initializer='he_normal'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x_train, y_train, epochs=3, verbose=0)

loss, acc = model.evaluate(x_test, y_test, verbose=0)
results['He'] = acc
print("Accuracy:", acc)

# -------- Xavier Initialization --------
print("\nXavier Initialization")

model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu', kernel_initializer='glorot_uniform'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x_train, y_train, epochs=3, verbose=0)

loss, acc = model.evaluate(x_test, y_test, verbose=0)
results['Xavier'] = acc
print("Accuracy:", acc)

# -------- Final Comparison --------
print("\nFinal Comparison:")
for k, v in results.items():
    print(f"{k} → {v:.4f}")
```
