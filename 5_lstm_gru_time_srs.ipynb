```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# Airline passenger dataset (monthly data)
data = np.array([
112,118,132,129,121,135,148,148,136,119,104,118,
115,126,141,135,125,149,170,170,158,133,114,140,
145,150,178,163,172,178,199,199,184,162,146,166
], dtype=float)

# Normalize
data = data / max(data)

# Create sequences
X, y = [], []
for i in range(len(data) - 5):
    X.append(data[i:i+5])
    y.append(data[i+5])

X = np.array(X)
y = np.array(y)

# Reshape for RNN
X = X.reshape(X.shape[0], X.shape[1], 1)

# Split
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# -------- LSTM --------
print("\nLSTM Model")

model = models.Sequential([
    layers.LSTM(32, input_shape=(5,1)),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')

model.fit(X_train, y_train, epochs=10, verbose=0)

lstm_loss = model.evaluate(X_test, y_test, verbose=0)
print("LSTM Loss:", lstm_loss)


# -------- GRU --------
print("\nGRU Model")

model = models.Sequential([
    layers.GRU(32, input_shape=(5,1)),
    layers.Dense(1)
])

model.compile(optimizer='adam', loss='mse')

model.fit(X_train, y_train, epochs=10, verbose=0)

gru_loss = model.evaluate(X_test, y_test, verbose=0)
print("GRU Loss:", gru_loss)
```
