```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.boston_housing.load_data()

# Normalize (important for regression)
mean = x_train.mean(axis=0)
std = x_train.std(axis=0)

x_train = (x_train - mean) / std
x_test = (x_test - mean) / std

results = {}

# -------- SGD --------
print("\nTraining with SGD optimizer")

model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(13,)),
    layers.Dense(1)   # Regression output
])

model.compile(optimizer='sgd',
              loss='mse',
              metrics=['mae'])

model.fit(x_train, y_train, epochs=50, verbose=0)

loss, mae = model.evaluate(x_test, y_test, verbose=0)
results['SGD'] = mae
print("SGD MAE:", mae)


# -------- ADAM --------
print("\nTraining with ADAM optimizer")

model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(13,)),
    layers.Dense(1)
])

model.compile(optimizer='adam',
              loss='mse',
              metrics=['mae'])

model.fit(x_train, y_train, epochs=50, verbose=0)

loss, mae = model.evaluate(x_test, y_test, verbose=0)
results['ADAM'] = mae
print("ADAM MAE:", mae)


# -------- RMSPROP --------
print("\nTraining with RMSPROP optimizer")

model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(13,)),
    layers.Dense(1)
])

model.compile(optimizer='rmsprop',
              loss='mse',
              metrics=['mae'])

model.fit(x_train, y_train, epochs=50, verbose=0)

loss, mae = model.evaluate(x_test, y_test, verbose=0)
results['RMSPROP'] = mae
print("RMSPROP MAE:", mae)


# -------- Final Comparison --------
print("\nFinal Comparison (Lower is better):")
for opt, mae in results.items():
    print(f"{opt} → MAE: {mae:.2f}")
```
